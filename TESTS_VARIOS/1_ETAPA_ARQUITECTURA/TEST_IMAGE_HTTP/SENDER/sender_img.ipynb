## Modo 1

```python
import ipywidgets as widgets
from IPython.display import display

uploader = widgets.FileUpload(
    accept='.jpg,.png,.jpeg',  # Extensiones permitidas
    multiple=False  # Solo un archivo a la vez
)

print("Selecciona una imagen:")
display(uploader)
```

```python
if uploader.value:
    # Obtener el primer archivo subido
    uploaded_file = uploader.value[0]
    filename = uploaded_file['name']
    content = uploaded_file['content'] # Contenido binario
    
    print(f"Archivo seleccionado: {filename}")
```

```python
import requests

url = "http://127.0.0.1:7000/upload/"
files = {"file": ("imgs/"+filename, content, "image/jpeg")}
response = requests.post(url, files=files)
print(response.json())                      
```

## Modo 2

```python
import requests

# Ruta de la imagen a enviar
image_path = "imgs/Selection_004.png"

# URL del servidor receptor
url = "http://127.0.0.1:7000/upload/"

# Leer la imagen en modo binario
# image_path es el nombre del archivo
# image_file es el contenido binario
# image/jpeg es el tipo MIME
with open(image_path, "rb") as image_file:
    files = {"file": (image_path, image_file, "image/jpeg")}
    
    # Enviar la imagen al servidor
    response = requests.post(url, files=files)

# Imprimir la respuesta del servidor
print(response.json())
```

## Modo 3

Se debe crear un servidor

```python
from fastapi import FastAPI, UploadFile, File
import requests

# Crear una instancia de FastAPI
app = FastAPI()

@app.post("/send-image/")
async def send_image(file: UploadFile = File(...)):
    """
    Endpoint para enviar una imagen a otro servidor receptor.
    """
    # URL del servidor receptor
    url = "http://127.0.0.1:8000/upload/"

    # Leer el archivo subido
    file_content = await file.read()
    files = {"file": (file.filename, file_content, file.content_type)}

    # Enviar la imagen al servidor receptor
    response = requests.post(url, files=files)
    return {"status_code": response.status_code, "response": response.json()}
```
